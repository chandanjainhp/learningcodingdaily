# lambda function


* Can have multiple parameters
* Must be written on a single line
* Have different syntax from regular functions
1. No return statement
2. No name

```python
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x ** 2, numbers))  # [1, 4, 9, 16]
print(squared)
points = [(2, 3), (1, 2), (4, 1)]
sorted_points = sorted(points, key=lambda p: p[1])  # Sort by second element
print(sorted_points)

```

# Generator in python

## yield

* you save memory
* you don't want the results immedietely
* lazly evalution

```python
def serve_chai():
    yield "Cup 1: Masala Chai"
    yield "Cup 1: Ginger Chai"
    yield "Cup 1: Elaichi Chai"

stall = serve_chai()

for cup in stall:
    print(cup)
```

```python
def get_chai_list():
    return ["Cup 1","Cup 2", "Cup 3"]

#generator function
def get_chai_gen():
    yield "Cup 1"
    yield "Cup 2"
    yield "Cup 3"

chai = get_chai_gen()
# print(f'chai {chai}' )
print(next(chai))
print(next(chai))
print(next(chai))

```

# infinite Generators

```python
def infinite_chai():
    count = 1
    while True:
        yield f"Refill #{count}"
        count +=1

user1 = infinite_chai()
user2 = infinite_chai()

for _ in range(3):
    u1=next(user1)
    print(f'user1 {u1}')
print(f'########## user yield is ended ###########')
for _ in range(3):
    u2=next(user2)
    print(f'user2 {u2}')
```

# yield from

# yeld close

```python
def local_chai():
    yield "Masala Chai"
    yield "Ginger Chai"

def imported_chai():
    yield "Matcha"
    yield "Oolong"

def full_menu():
    yield from local_chai()
    yield from imported_chai()

for chai in full_menu():
    print(chai)


```

```python
def chai_stall():
    try:
        while True:
            order = yield "Waiting for chai order"
    except:
        print("Stall closed, No more chai")


stall = chai_stall()
print(next(stall))
stall.close() #cleanup
```

```python
import random

class Result:
    @staticmethod
    def countGreaterThanK(arr, k):
        """
        Counts elements strictly greater than k.

        Args:
            arr (list[int]): List of integers
            k (int): Threshold value

        Returns:
            int: Count of elements greater than k
        """
        # Write your code here
        counter=0
        for i in range(0,len(arr)):
            if arr[i] > k:
                counter+=1
        return counter   
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 20) for _ in range(n)]
    k = random.randint(1, 10)

    print("Array:", arr)
    print("k:", k)

    result = Result.countGreaterThanK(arr, k)
    print("Count > k:", result)
```
