```python
import random

class Result:
    @staticmethod
    def sumAllElements(arr):
        """
        Calculates the sum of all elements in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: Sum of all elements
        """
        # Write your code here
        n=len(arr)
        sumArr = 0
        for i in range(0,n):
            sumArr+=arr[i]
        return sumArr
        
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-20, 20) for _ in range(n)]

    print("Array:", arr)

    result = Result.sumAllElements(arr)
    print("Sum of Elements:", result)
```

```python
import random

class Result:
    @staticmethod
    def findAverage(arr):
        """
        Finds the average of the array elements.

        Args:
            arr (list[int]): List of integers

        Returns:
            float: Average of elements
        """
        # Write your code here
        SubArr=0
        AvgSubarr=0
        for i in range(0,len(arr)):
            SubArr+=arr[i]
            AvgSubarr=SubArr//2
        return AvgSubarr
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-20, 20) for _ in range(n)]

    print("Array:", arr)

    result = Result.findAverage(arr)
    print("Average:", result)
```

```python
import random

class Result:
    @staticmethod
    def maxDifference(arr):
        """
        Finds the maximum difference arr[j] - arr[i] such that j > i.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: Maximum difference or 0 if none exists
        """
        # Write your code here
        if len(arr) < 2:
            return 0

        maxProfit = 0
        minPrice = float("inf")
        n = len(arr)
        for i in range(0,n):
            minPrice = min(minPrice, arr[i])
            maxProfit = max(maxProfit,arr[i]-minPrice)
        return maxProfit   
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-20, 20) for _ in range(n)]

    print("Array:", arr)

    result = Result.maxDifference(arr)
    print("Maximum Difference:", result)
```
