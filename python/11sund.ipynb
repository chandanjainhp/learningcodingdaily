```python
import random

class Result:
    @staticmethod
    def longestSubarraySumAtMostK(arr, k):
        """
        Finds the length of the longest subarray with sum <= k.

        Args:
            arr (list[int]): List of positive integers
            k (int): Maximum allowed sum

        Returns:
            int: Length of the longest valid subarray
        """
        l = 0
        curr_sum = 0
        max_len = 0

        for r in range(len(arr)):
            curr_sum += arr[r]

            while curr_sum > k:
                curr_sum -= arr[l]
                l += 1

            max_len = max(max_len, r - l + 1)

        return max_len


if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 10) for _ in range(n)]
    k = random.randint(5, 30)

    print("Array:", arr)
    print("k:", k)

    result = Result.longestSubarraySumAtMostK(arr, k)
    print("Longest Subarray Length:", result)
```

```python
import random

class Result:
    @staticmethod
    def countOccurrences(arr, target):
        """
        Counts how many times target appears in the array.

        Args:
            arr (list[int]): List of integers
            target (int): Value to count

        Returns:
            int: Number of occurrences of target
        """
        # Write your code here
        count=0
        for i in range(0,len(arr)):
            if arr[i] == target:
                count+=1
        return count
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 10) for _ in range(n)]
    target = random.randint(1, 10)

    print("Array:", arr)
    print("Target:", target)

    result = Result.countOccurrences(arr, target)
    print("Target Count:", result)
```

```python
import random

class Result:
    @staticmethod
    def findMinElement(arr):
        """
        Finds the minimum element in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: Minimum value in the array
        """
        # Write your code here
        minlenght =0
        for nums in arr:
            if nums < minlenght:
                minlenght = nums
        return minlenght
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-50, 50) for _ in range(n)]

    print("Array:", arr)

    result = Result.findMinElement(arr)
    print("Minimum Element:", result)
```

```python
import random

class Result:
    @staticmethod
    def removeDuplicates(arr):
        """
        Removes duplicates from the array while preserving order.

        Args:
            arr (list[int]): List of integers

        Returns:
            list[int]: Array with duplicates removed
        """
        # Write your code here
        numslist = []
        for nums in arr:
            if nums not in numslist:
                numslist.append(nums)
        return numslist
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 10) for _ in range(n)]

    print("Array:", arr)

    result = Result.removeDuplicates(arr)
    print("Array After Removing Duplicates:", result)
```
