```python
import random

class Result:
    @staticmethod
    def reverseArray(arr):
        """
        Reverses the given array.

        Args:
            arr (list[int]): List of integers

        Returns:
            list[int]: Reversed array
        """
        # Write your code here
        right = len(arr)-1
        left = 0
        while left < right:
             arr[left],arr[ right] = arr[ right],arr[left]
             left+=1
             right-=1
        return arr
if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-50, 50) for _ in range(n)]

    print("Original Array:", arr)

    result = Result.reverseArray(arr)
    print("Reversed Array:", result)
```

```python
import random

class Result:
    @staticmethod
    def isSorted(arr):
        """
        Checks if the array is sorted in non-decreasing order.

        Args:
            arr (list[int]): List of integers

        Returns:
            bool: True if sorted, False otherwise
        """
        n = len(arr)
        for i in range(n - 1):
            if arr[i] > arr[i + 1]:
                return False
        return True


if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-20, 20) for _ in range(n)]

    print("Array:", arr)
    result = Result.isSorted(arr)
    print("Is Sorted:", result)
```

```python
def reverse(nums, left, right):
    while left < right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1


def rotate(nums, k):
    n = len(nums)
    k = k % n  # handle k > n

    # Step 1: reverse last k elements
    reverse(nums, n - k, n - 1)

    # Step 2: reverse first n-k elements
    reverse(nums, 0, n - k - 1)

    # Step 3: reverse whole array
    reverse(nums, 0, n - 1)


if __name__ == "__main__":
    nums = [3, 9, 5, 6, 7, 2, 10, 9]
    k = 5

    print("Original array:", nums)
    rotate(nums, k)
    print("Array after rotating right by", k, ":", nums)
```

```python
def max_points_from_cards(nums, k):
    n = len(nums)

    # Take first k cards from the left
    left_sum = 0
    for i in range(k):
        left_sum += nums[i]

    max_sum = left_sum

    # Take cards from the right one by one
    right_sum = 0
    right_index = n - 1

    for i in range(k - 1, -1, -1):
        left_sum -= nums[i]
        right_sum += nums[right_index]
        right_index -= 1

        max_sum = max(max_sum, left_sum + right_sum)

    return max_sum


if __name__ == "__main__":
    # Input
    nums = [6, 2, 3, 4, 7, 2, 1, 7, 1]
    k = 4

    # Function call
    result = max_points_from_cards(nums, k)

    # Output
    print("Cards:", nums)
    print("K:", k)
    print("Maximum points you can obtain:", result)
```

```python
from typing import List

class Solution:
    def maxFrequency(self, nums: List[int], k: int) -> int:
        nums.sort()

        l = 0
        total = 0
        res = 0

        for r in range(len(nums)):
            total += nums[r]

            # shrink window if cost exceeds k
            while nums[r] * (r - l + 1) > total + k:
                total -= nums[l]
                l += 1

            res = max(res, r - l + 1)

        return res


if __name__ == "__main__":
    # Example input
    nums = [1, 2, 4]
    k = 5

    sol = Solution()
    result = sol.maxFrequency(nums, k)

    print("Input array:", nums)
    print("k:", k)
    print("Maximum frequency:", result)
```
