```python
import random

class Result:
    @staticmethod
    def countEvenNumbers(arr):
        """
        Counts the number of even elements in the array.

        Args:
            arr (List[int]): List of integers

        Returns:
            int: Count of even numbers
        """
        counter = 0
        for i in arr:
            if i%2 == 0:
                counter+=1
        return counter 
        pass

if __name__ == "__main__":
    n = random.randint(5, 15)
    arr = [random.randint(-50, 50) for _ in range(n)]

    print("Array:", arr)

    result = Result.countEvenNumbers(arr)
    print("Even Count:", result)
```

```python
import random
import string

class Result:
    @staticmethod
    def reverseString(s):
        """
        Reverses the given string.

        Args:
            s (str): Input string

        Returns:
            str: Reversed string
        """
        string = ''
        for i in s:
            string = i + string
        return string
        pass

if __name__ == "__main__":
    s = ''.join(random.choices(string.ascii_lowercase, k=random.randint(5, 15)))

    print("String:", s)

    result = Result.reverseString(s)
    print("Reversed String:", result)
```

```python
import random

class Result:
    @staticmethod
    def findPairWithSum(arr, target):
        """
        Checks if any pair sums to the target value using brute force.

        Args:
            arr (List[int]): List of integers
            target (int): Target sum

        Returns:
            bool: True if pair exists, else False
        """
        n = len(arr)

        for i in range(n):
            for j in range(i + 1, n):
                if arr[i] + arr[j] == target:
                    return arr[i] , arr[j]

        return False


if __name__ == "__main__":
    n = random.randint(5, 15)
    arr = [random.randint(-50, 50) for _ in range(n)]
    target = random.randint(-20, 20)

    print("Array:", arr)
    print("Target:", target)

    result = Result.findPairWithSum(arr, target)
    print("Pair Exists:", result)
```

```python
import random

class Result:
    @staticmethod
    def maxSubarraySum(arr, k):
        """
        Finds maximum sum of a contiguous subarray of size k.

        Args:
            arr (List[int]): List of integers
            k (int): Subarray size

        Returns:
            int: Maximum subarray sum
        """
        n = len(arr)
        # subarray=0
        maxsubarray = float('-inf')
        for i in range(n-k+1):
            subarray=0
            for j in range(i,k+i):
                subarray+=arr[i]
                maxsubarray=max(maxsubarray,subarray)
        return maxsubarray      
        pass

if __name__ == "__main__":
    n = random.randint(5, 15)
    k = random.randint(1, n)
    arr = [random.randint(-50, 50) for _ in range(n)]

    print("Array:", arr)
    print("k:", k)

    result = Result.maxSubarraySum(arr, k)
    print("Maximum Sum:", result)
```

```python
# Python program to find the length of the longest
# subarray having sum k using nested loop

# Function to find longest sub-array having sum k
def longestSubarray(arr, k):
    res = 0

    for i in range(len(arr)):
        
        # Sum of subarray from i to j
        sum = 0
        for j in range(i, len(arr)):
            sum += arr[j]
          
            # If subarray sum is equal to k
            if sum == k:
              
                # find subarray length and update result
                subLen = j - i + 1
                res = max(res, subLen)
    
    return res

if __name__ == "__main__":
    arr = [10, 5, 2, 7, 1, -10]
    k = 15
    print(longestSubarray(arr, k))
```

```python
def longestSubarray(arr, k):
    n = len(arr)
    l = 0
    r = 0
    curr_sum = 0
    max_len = 0

    while r < n:
        curr_sum += arr[r]

        while curr_sum > k:
            curr_sum -= arr[l]
            l += 1
            if curr_sum <= k:
                max_len = max(max_len, r - l + 1)
                r += 1

    return max_len


if __name__ == "__main__":
    arr = [2, 5, 1, 7, 10]
    k = 15
    print(longestSubarray(arr, k))
```
