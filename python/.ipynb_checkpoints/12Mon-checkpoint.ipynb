# lambda function


* Can have multiple parameters
* Must be written on a single line
* Have different syntax from regular functions
1. No return statement
2. No name

```python
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x ** 2, numbers))  # [1, 4, 9, 16]
print(squared)
points = [(2, 3), (1, 2), (4, 1)]
sorted_points = sorted(points, key=lambda p: p[1])  # Sort by second element
print(sorted_points)

```

```python


def squares(num):
    return num**2
def cubes(num):
    return num**3


def transform_list(num_list,transform_item):
    transformed_0 = transform_item(num_list[0])
    transformed_1 = transform_item(num_list[1])
    return [transformed_0,transformed_1]



transform_list([2,3],squares)



```

```python
transform_list([2,3],lambda num:num **3)
```

```python
array = [2,34,5,6]

list(map(lambda num:num**3,array))
```

```python
array = [2,34,5,6]

list(filter(lambda num:num%2==0, array))
```

# Generator in python

## yield

* you save memory
* you don't want the results immedietely
* lazly evalution

```python
def serve_chai():
    yield "Cup 1: Masala Chai"
    yield "Cup 1: Ginger Chai"
    yield "Cup 1: Elaichi Chai"

stall = serve_chai()

for cup in stall:
    print(cup)
```

```python
def get_chai_list():
    return ["Cup 1","Cup 2", "Cup 3"]

#generator function
def get_chai_gen():
    yield "Cup 1"
    yield "Cup 2"
    yield "Cup 3"

chai = get_chai_gen()
# print(f'chai {chai}' )
print(next(chai))
print(next(chai))
print(next(chai))

```

# infinite Generators

```python
def infinite_chai():
    count = 1
    while True:
        yield f"Refill #{count}"
        count +=1

user1 = infinite_chai()
user2 = infinite_chai()

for _ in range(3):
    u1=next(user1)
    print(f'user1 {u1}')
print(f'########## user yield is ended ###########')
for _ in range(3):
    u2=next(user2)
    print(f'user2 {u2}')
```

## Send value to Generator

```python
def chai_customer():
    print("welcome ! what Chai would you like?")
    order = yield
    while True:
        print(f"Preparing {order}")
        order = yield
stall = chai_customer()
next(stall)

stall.send("Masala chai")
```

this allows stop the programing order = yield


# yield from

# yeld close

```python
def local_chai():
    yield "Masala Chai"
    yield "Ginger Chai"

def imported_chai():
    yield "Matcha"
    yield "Oolong"

def full_menu():
    yield from local_chai()
    yield from imported_chai()

for chai in full_menu():
    print(chai)


```

```python
def chai_stall():
    try:
        while True:
            order = yield "Waiting for chai order"
    except:
        print("Stall closed, No more chai")


stall = chai_stall()
print(next(stall))
stall.close() #cleanup
```

Constructor

```python
class BankAccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def show_details(self):
        return print(f'The account holder name: {self.account_holder}, your balance in account: {self.balance}')


user1 = BankAccount("chandan jain H P", 20000)
user1.show_details()


```
