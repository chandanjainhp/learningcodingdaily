```python
# Given arrays
n = [5, 3, 2, 2, 1, 5, 5, 7, 5, 10]
m = [10, 11, 1, 9, 5, 67, 2]

# Constraints
MAX_VALUE = 10

# Step 1: Create hash list (frequency array)
hash_list = [0] * (MAX_VALUE + 1)
# Step 2: Fill hash list using array n
for num in n:
    hash_list[num] += 1

# Step 3: Answer queries from m
for num in m:
    if num < 1 or num > MAX_VALUE:
        print(0)
    else:
        print(f"{hash_list[num]}")
```

```python
# Given string and queries
s = "azyayyzzaaaa"
q = ["a", "a", "y", "z"]

# Step 1: Create hash list for 26 lowercase letters
hash_list = [0] * 26

# Step 2: Fill hash list using string s
print("Building frequency table:")
for ch in s:
    asciival = ord(ch)  # ASCII value: 97-122 for 'a'-'z'
    index = asciival - 97  # Convert to 0-25 index
    print(f"  Character '{ch}' -> ASCII {asciival} -> Index {index}")
    hash_list[index] += 1

# Display the frequency table
print("\nFrequency table:")
for i in range(26):
    if hash_list[i] > 0:
        char = chr(i + 97)
        print(f"  '{char}': {hash_list[i]}")

# Step 3: Answer queries
print("\nQuery results:")
for ch in q:
    asciival = ord(ch)
    index = asciival - 97
    count = hash_list[index]
    print(f"  Character '{ch}' appears {count} times")

# '''
# The ord() function returns the number representing the unicode code of a specified character.
# ASCII Reference:
# - Lowercase letters: 97-122 ('a' to 'z')
# - Uppercase letters: 65-90 ('A' to 'Z')
# - https://theasciicode.com.ar/ascii-printable-characters/lowercase-letter-b-minuscule-ascii-code-98.html
# '''

# Alternative: Using Python's Counter (more Pythonic)
# from collections import Counter

# print("\n--- Alternative using Counter ---")
# freq = Counter(s)
# for ch in q:
#     print(f"  Character '{ch}' appears {freq[ch]} times")
```

```python
num = [5, 3, 2, 2, 1, 5, 5, 7, 5, 10]

dis= dict()
n=len(num)
for i in range(0,n):
    if num[i] in dis:
        dis[num[i]]+=1
    else:
        dis[num[i]]=1
print(dis)    
    
```
