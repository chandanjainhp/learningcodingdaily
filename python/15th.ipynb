```python
import random

class Result:
    @staticmethod
    def elementExists(arr, target):
        """
        Checks whether target exists in the array.

        Args:
            arr (list[int]): List of integers
            target (int): Element to search for

        Returns:
            bool: True if target exists, False otherwise
        """
        # Write your code here
        n = len(arr)
        for i in range(0,n):
            if arr[i] == target:
                return True
        return False
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 20) for _ in range(n)]
    target = random.randint(1, 20)

    print("Array:", arr)
    print("Target:", target)

    result = Result.elementExists(arr, target)
    print("Exists:", result)
```

```python
import random

class Result:
    @staticmethod
    def countOddNumbers(arr):
        """
        Counts the number of odd integers in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: Count of odd numbers
        """
        # Write your code here
        count=0
        n = len(arr)
        for i in range(0,n):
            if arr[i]%2==1:
                count+=1
        return count
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-20, 20) for _ in range(n)]

    print("Array:", arr)

    result = Result.countOddNumbers(arr)
    print("Odd Numbers Count:", result)
```

```python
import random

class Result:
    @staticmethod
    def majorityElement(arr):
        """
        Finds the majority element in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: Majority element or -1 if none exists
        """
        # Step 1: Find candidate
        count = 0
        candidate = None

        for num in arr:
            if count == 0:
                candidate = num
                count = 1
            elif num == candidate:
                count += 1
            else:
                count -= 1

        # Step 2: Verify candidate
        if arr.count(candidate) > len(arr) // 2:
            return candidate

        return -1

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 5) for _ in range(n)]

    print("Array:", arr)

    result = Result.majorityElement(arr)
    print("Majority Element:", result)
```
