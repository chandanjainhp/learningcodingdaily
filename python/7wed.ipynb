```python
def longest_subarray_sum_le_k(arr, k):
    n = len(arr)
    l = 0
    r = 0
    curr_sum = 0
    max_len = 0

    while r < n:
        curr_sum += arr[r]

        while curr_sum > k:
            curr_sum -= arr[l]
            l += 1

        if curr_sum <= k:
            max_len = max(max_len, r - l + 1)

        r += 1

    return max_len


if __name__ == "__main__":
    arr = [2, 5, 1, 10, 10]
    k = 14

    result = longest_subarray_sum_le_k(arr, k)

    print("Array:", arr)
    print("K:", k)
    print("Longest subarray length with sum ≤ K:", result)
```

```python
from typing import List

class Solution:
    def maxFrequency(self, nums: List[int], k: int) -> int:
        nums.sort()

        l = 0
        total = 0
        res = 0

        for r in range(len(nums)):
            total += nums[r]
            # window size of the r - l + 1
            # shrink window if cost exceeds k
            while nums[r] * (r - l + 1) > total + k:
                total -= nums[l]
                l += 1

            res = max(res, r - l + 1)

        return res


if __name__ == "__main__":
    # Example input
    nums = [1, 2, 4]
    k = 5

    sol = Solution()
    result = sol.maxFrequency(nums, k)

    print("Input array:", nums)
    print("k:", k)
    print("Maximum frequency:", result)
```
