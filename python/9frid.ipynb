```python
import random

class Result:
    @staticmethod
    def countPositiveNumbers(arr):
        """
        Counts the number of positive integers in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: Count of positive numbers
        """
        count = 0
        for i in range(0,len(arr)):
            if arr[i] >= 0:
                count+=1
        return count
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-10, 10) for _ in range(n)]

    print("Array:", arr)

    result = Result.countPositiveNumbers(arr)
    print("Positive Count:", result)
```

```python
import random

class Result:
    @staticmethod
    def findMaxElement(arr):
        """
        Finds the maximum element in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: Maximum value in the array
        """
        maxElem = 0
        for i in range(0,len(arr)):
            if arr[i] > maxElem:
                maxElem =arr[i]
        return maxElem
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-50, 50) for _ in range(n)]

    print("Array:", arr)

    result = Result.findMaxElement(arr)
    print("Maximum Element:", result)
```

```python
import random

class Result:
    @staticmethod
    def countDistinctElements(arr):
        """
        Counts the number of distinct elements in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: Number of unique elements
        """
        frequmentElement = dict()

        for i in range(0,len(arr)):
            if arr[i] in frequmentElement:
                frequmentElement[arr[i]]+=1
            else:
                frequmentElement[arr[i]]=1
        return frequmentElement
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 10) for _ in range(n)]

    print("Array:", arr)

    result = Result.countDistinctElements(arr)
    print("Distinct Elements Count:", result)
```

```python
import random

class Result:
    @staticmethod
    def firstNonRepeating(arr):
        """
        Finds the first non-repeating element in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: First non-repeating element or -1 if none exists
        """
        freq = {}

        # Step 1: Count frequency of each element
        for num in arr:
            freq[num] = freq.get(num, 0) + 1

        # Step 2: Find first element with frequency 1
        for num in arr:
            if freq[num] == 1:
                return num

        return -1


if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 10) for _ in range(n)]

    print("Array:", arr)

    result = Result.firstNonRepeating(arr)
    print("First Non-Repeating Element:", result)
```
