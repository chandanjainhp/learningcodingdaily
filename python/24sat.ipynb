```python
import random

class Result:
    @staticmethod
    def firstDuplicate(arr):
        """
        Finds the first duplicate element in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: First duplicate element or -1 if none exists
        """
        # Write your code here

        firstDuplation = dict{}
        for i in arr:
            if arr[i] in firstDuplation:
                firstDuplation[arr[i]] +=1
            else 
                firstDuplation[arr[i]] =1
        
      
        return -1       
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 10) for _ in range(n)]

    print("Array:", arr)

    result = Result.firstDuplicate(arr)
    print("First Duplicate:", result)
```

```python
import random

class Result:
    @staticmethod
    def countFrequency(arr):
        """
        Counts frequency of each element in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            dict[int, int]: Frequency of each element
        """
        countFrequency = dict()
        # Write your code here
        for i in range(0,len(arr)):
            if arr[i] in countFrequency:
                countFrequency[arr[i]] +=1
            else :
                countFrequency[arr[i]] =1
        return countFrequency        
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 10) for _ in range(n)]

    print("Array:", arr)

    result = Result.countFrequency(arr)
    print("Frequencies:", result)
```

```python
import random

class Result:
    @staticmethod
    def areArraysEqual(arr1, arr2):
        """
        Checks if two arrays are equal ignoring order.
        
        Args:
            arr1 (list[int]): First array
            arr2 (list[int]): Second array
        
        Returns:
            bool: True if equal ignoring order, False otherwise
        """
        # Check if lengths are different
        if len(arr1) != len(arr2):
            return False
        
        frequency1 = dict()
        frequency2 = dict()
        
        # Build frequency map for arr1
        for i in range(len(arr1)):
            if arr1[i] in frequency1:
                frequency1[arr1[i]] += 1
            else:
                frequency1[arr1[i]] = 1
        
        # Build frequency map for arr2
        for i in range(len(arr2)):
            if arr2[i] in frequency2:
                frequency2[arr2[i]] += 1
            else:
                frequency2[arr2[i]] = 1
        
        # Compare the two frequency maps
        return frequency1 == frequency2


if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr1 = [random.randint(1, 10) for _ in range(n)]
    arr2 = arr1.copy()
    random.shuffle(arr2)
    
    print("Array 1:", arr1)
    print("Array 2:", arr2)
    result = Result.areArraysEqual(arr1, arr2)
    print("Are Equal:", result)
    
    # Additional test cases
    print("\nTest with different arrays:")
    arr3 = [1, 2, 3]
    arr4 = [1, 2, 4]
    print("Array 3:", arr3)
    print("Array 4:", arr4)
    print("Are Equal:", Result.areArraysEqual(arr3, arr4))
```

```python
import random

class Result:
    @staticmethod
    def longestZeroSumSubarray(arr):
        """
        Finds the length of the longest subarray with sum zero.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: Length of the longest zero-sum subarray
        """
        # Write your code here

        n = len(arr)
        max_len = 0
        curr_sum = 0
        sum_map = {}  # Maps cumulative sum to its first occurrence index
        
        for i in range(n):
            curr_sum += arr[i]
            
            # If cumulative sum is 0, subarray from index 0 to i has sum 0
            if curr_sum == 0:
                max_len = i + 1
            
            # If we've seen this cumulative sum before, the subarray between
            # the previous occurrence and current index has sum 0
            if curr_sum in sum_map:
                max_len = max(max_len, i - sum_map[curr_sum])
            else:
                # Store first occurrence of this cumulative sum
                sum_map[curr_sum] = i
        
        return max_len
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-5, 5) for _ in range(n)]

    print("Array:", arr)

    result = Result.longestZeroSumSubarray(arr)
    print("Longest Zero-Sum Subarray Length:", result)
```
