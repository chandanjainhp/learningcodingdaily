```python
import random

class Result:
    @staticmethod
    def rotate(arr, left, right):
        while left < right:
            arr[left], arr[right] = arr[right], arr[left]
            left += 1
            right -= 1

    @staticmethod
    def rotateRight(arr, k):
        """
        Rotates the array to the right by k steps.

        Args:
            arr (list[int]): List of integers
            k (int): Number of rotations

        Returns:
            list[int]: Rotated array
        """
        n = len(arr)
        if n == 0:
            return arr

        k = k % n

        # Step 1: Reverse entire array
        Result.rotate(arr, 0, n - 1)

        # Step 2: Reverse first k elements
        Result.rotate(arr, 0, k - 1)

        # Step 3: Reverse remaining elements
        Result.rotate(arr, k, n - 1)

        return arr

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 20) for _ in range(n)]
    k = random.randint(0, 20)

    print("Array:", arr)
    print("k:", k)

    result = Result.rotateRight(arr, k)
    print("Rotated Array:", result)
```

```python
import random

class Result:
    @staticmethod
    def minSubArrayLen(target, arr):
        """
        Finds the minimum length of a subarray with sum >= target.
        Args:
            target (int): Target sum
            arr (list[int]): List of positive integers
        Returns:
            int: Minimum subarray length or 0 if none exists
        """
        n = len(arr)
        minLen = float('inf')  # Start with infinity
        
        # Check all possible subarrays
        for i in range(n):
            subArr = 0
            for j in range(i, n):
                subArr += arr[j]
                if subArr >= target:
                    # Calculate length: j - i + 1
                    minLen = min(minLen, j - i + 1)
                    break  # No need to continue, we found valid subarray starting at i
        
        # Return 0 if no valid subarray found, otherwise return minLen
        return 0 if minLen == float('inf') else minLen

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 10) for _ in range(n)]
    target = random.randint(5, 30)
    print("Array:", arr)
    print("Target:", target)
    result = Result.minSubArrayLen(target, arr)
    print("Minimum Subarray Length:", result)
```
