```python
import random

class Result:
    @staticmethod
    def removeDuplicates(arr):
        """
        Removes duplicates from a sorted array in place.

        Args:
            arr (list[int]): Sorted list of integers

        Returns:
            int: New length after removing duplicates
        """
        # Write your code here
        if not arr:
            return 0

        slow = 0

        for fast in range(1, len(arr)):
            if arr[fast] != arr[slow]:
                slow += 1
                arr[slow] = arr[fast]

        return slow + 1

if __name__ == "__main__":
    # Random test data generation (for local testing)
    arr = sorted([random.randint(1, 5) for _ in range(random.randint(5, 15))])

    print("Original Array:", arr)

    length = Result.removeDuplicates(arr)
    print("New Length:", length)
    print("Modified Array:", arr[:length])
```
