```python
import random

class Result:
    @staticmethod
    def countConsecutiveEquals(arr):
        """
        Counts how many times consecutive elements are equal.

        Args:
            arr (list[int]): List of integers

        Returns:
            int: Count of consecutive equal pairs
        """
        count = 0

        for i in range(len(arr) - 1):
            if arr[i] == arr[i + 1]:
                count += 1

        return count


if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 5) for _ in range(n)]

    print("Array:", arr)

    result = Result.countConsecutiveEquals(arr)
    print("Consecutive Equal Count:", result)

```

```python
import random

class Result:
    @staticmethod
    def isPalindromeArray(arr):
        """
        Checks if the array is a palindrome.

        Args:
            arr (list[int]): List of integers

        Returns:
            bool: True if palindrome, False otherwise
        """
        # Write your code here
        left = 0
        right = len(arr)-1

        while left < right:
            if arr[left] == arr[right]:
                return False
            left+=1
            right-=1
        return True
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 5) for _ in range(n)]

    print("Array:", arr)

    result = Result.isPalindromeArray(arr)
    print("Is Palindrome:", result)
```

```python
import random

class Result:
    @staticmethod
    def findLeaders(arr):
        """
        Finds all leader elements in the array.

        Args:
            arr (list[int]): List of integers

        Returns:
            list[int]: List of leaders in original order
        """
        # Write your code here
        leaders = []
        for i  in range(0,len(arr)):
            maxnum= max(arr)
            if maxnum == arr[i]:
                leaders.append(maxnum)
        
        return leaders
        pass

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 20) for _ in range(n)]

    print("Array:", arr)

    result = Result.findLeaders(arr)
    print("Leaders:", result)
```

```python
import random

class Result:
    @staticmethod
    def countSubarraysWithSum(arr, k):
        """
        Counts the number of contiguous subarrays with sum equal to k.
        Args:
            arr (list[int]): List of integers
            k (int): Target sum
        Returns:
            int: Number of subarrays with sum k
        """
        # Two-pointer approach (only works for non-negative arrays)
        n = len(arr)
        count = 0
        
        for i in range(n):
            curr_sum = 0
            for j in range(i, n):
                curr_sum += arr[j]
                
                if curr_sum == k:
                    count += 1
                elif curr_sum > k:
                    break  # No point continuing if sum exceeds k (only for non-negative)
        
        return count

if __name__ == "__main__":
    # Random test data generation (for local testing)
    # Modified to use only non-negative numbers for two-pointer approach
    n = random.randint(5, 15)
    arr = [random.randint(0, 5) for _ in range(n)]  # Changed to 0-5 (non-negative)
    k = random.randint(0, 10)  # Changed to non-negative
    print("Array:", arr)
    print("Target Sum k:", k)
    result = Result.countSubarraysWithSum(arr, k)
    print("Subarrays Count:", result)
```
