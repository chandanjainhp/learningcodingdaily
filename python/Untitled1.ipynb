```python
# ============================================
# 1. BUBBLE SORT - O(n²)
# ============================================
def bubble_sort(arr, n):
    """Simple swapping of adjacent elements"""
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                # Swap
                temp = arr[j]
                arr[j] = arr[j + 1]
                arr[j + 1] = temp
    return arr


# ============================================
# 2. SELECTION SORT - O(n²)
# ============================================
def selection_sort(arr, n):
    """Find minimum and place it at beginning"""
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        
        # Swap minimum with first element
        temp = arr[i]
        arr[i] = arr[min_idx]
        arr[min_idx] = temp
    
    return arr


# ============================================
# 3. INSERTION SORT - O(n²), good for nearly sorted
# ============================================
def insertion_sort(arr, n):
    """Insert each element in its correct position"""
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        
        # Move elements greater than key one position ahead
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j = j - 1
        
        arr[j + 1] = key
    
    return arr


# ============================================
# 4. MERGE SORT - O(n log n), stable
# ============================================
def merge(arr, left, mid, right):
    """Merge two sorted subarrays"""
    n1 = mid - left + 1
    n2 = right - mid
    
    # Create temp arrays
    L = [0] * n1
    R = [0] * n2
    
    # Copy data to temp arrays
    for i in range(n1):
        L[i] = arr[left + i]
    for j in range(n2):
        R[j] = arr[mid + 1 + j]
    
    # Merge temp arrays back
    i = 0  # Initial index of first subarray
    j = 0  # Initial index of second subarray
    k = left  # Initial index of merged subarray
    
    while i < n1 and j < n2:
        if L[i] <= R[j]:
            arr[k] = L[i]
            i = i + 1
        else:
            arr[k] = R[j]
            j = j + 1
        k = k + 1
    
    # Copy remaining elements
    while i < n1:
        arr[k] = L[i]
        i = i + 1
        k = k + 1
    
    while j < n2:
        arr[k] = R[j]
        j = j + 1
        k = k + 1


def merge_sort(arr, left, right):
    """Divide and conquer sorting"""
    if left < right:
        mid = left + (right - left) // 2
        
        # Sort first and second halves
        merge_sort(arr, left, mid)
        merge_sort(arr, mid + 1, right)
        merge(arr, left, mid, right)
    
    return arr


# ============================================
# 5. QUICK SORT - O(n log n) average
# ============================================
def partition(arr, low, high):
    """Partition array around pivot"""
    pivot = arr[high]  # Choose last element as pivot
    i = low - 1  # Index of smaller element
    
    for j in range(low, high):
        if arr[j] <= pivot:
            i = i + 1
            # Swap
            temp = arr[i]
            arr[i] = arr[j]
            arr[j] = temp
    
    # Swap pivot to correct position
    temp = arr[i + 1]
    arr[i + 1] = arr[high]
    arr[high] = temp
    
    return i + 1


def quick_sort(arr, low, high):
    """Divide and conquer with partitioning"""
    if low < high:
        # Partitioning index
        pi = partition(arr, low, high)
        
        # Recursively sort elements before and after partition
        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)
    
    return arr


# ============================================
# 6. HEAP SORT - O(n log n)
# ============================================
def heapify(arr, n, i):
    """Heapify subtree rooted at index i"""
    largest = i  # Initialize largest as root
    left = 2 * i + 1
    right = 2 * i + 2
    
    # If left child exists and is greater than root
    if left < n and arr[left] > arr[largest]:
        largest = left
    
    # If right child exists and is greater than largest
    if right < n and arr[right] > arr[largest]:
        largest = right
    
    # If largest is not root
    if largest != i:
        # Swap
        temp = arr[i]
        arr[i] = arr[largest]
        arr[largest] = temp
        
        # Recursively heapify the affected sub-tree
        heapify(arr, n, largest)


def heap_sort(arr, n):
    """Build max heap and extract elements"""
    # Build max heap
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)
    
    # Extract elements from heap one by one
    for i in range(n - 1, 0, -1):
        # Swap current root to end
        temp = arr[0]
        arr[0] = arr[i]
        arr[i] = temp
        
        # Heapify the reduced heap
        heapify(arr, i, 0)
    
    return arr


# ============================================
# 7. COUNTING SORT - O(n + k), for integers
# ============================================
def counting_sort(arr, n):
    """Sort by counting occurrences"""
    # Find maximum element
    max_val = arr[0]
    for i in range(1, n):
        if arr[i] > max_val:
            max_val = arr[i]
    
    # Create count array
    count = [0] * (max_val + 1)
    output = [0] * n
    
    # Store count of each element
    for i in range(n):
        count[arr[i]] = count[arr[i]] + 1
    
    # Change count[i] to contain actual position
    for i in range(1, max_val + 1):
        count[i] = count[i] + count[i - 1]
    
    # Build output array
    for i in range(n - 1, -1, -1):
        output[count[arr[i]] - 1] = arr[i]
        count[arr[i]] = count[arr[i]] - 1
    
    # Copy output array to arr
    for i in range(n):
        arr[i] = output[i]
    
    return arr


# ============================================
# 8. RADIX SORT - O(d * (n + k))
# ============================================
def counting_sort_for_radix(arr, n, exp):
    """Counting sort based on digit represented by exp"""
    output = [0] * n
    count = [0] * 10
    
    # Store count of occurrences
    for i in range(n):
        index = arr[i] // exp
        count[index % 10] = count[index % 10] + 1
    
    # Change count[i] to contain actual position
    for i in range(1, 10):
        count[i] = count[i] + count[i - 1]
    
    # Build output array
    for i in range(n - 1, -1, -1):
        index = arr[i] // exp
        output[count[index % 10] - 1] = arr[i]
        count[index % 10] = count[index % 10] - 1
    
    # Copy output to arr
    for i in range(n):
        arr[i] = output[i]


def radix_sort(arr, n):
    """Sort by processing digits from least to most significant"""
    # Find maximum number to know number of digits
    max_val = arr[0]
    for i in range(1, n):
        if arr[i] > max_val:
            max_val = arr[i]
    
    # Do counting sort for every digit
    exp = 1
    while max_val // exp > 0:
        counting_sort_for_radix(arr, n, exp)
        exp = exp * 10
    
    return arr


# ============================================
# MAIN FUNCTION - Test all sorts
# ============================================
def main():
    # Test array
    arr1 = [64, 34, 25, 12, 22, 11, 90]
    arr2 = [64, 34, 25, 12, 22, 11, 90]
    arr3 = [64, 34, 25, 12, 22, 11, 90]
    arr4 = [64, 34, 25, 12, 22, 11, 90]
    arr5 = [64, 34, 25, 12, 22, 11, 90]
    arr6 = [64, 34, 25, 12, 22, 11, 90]
    arr7 = [64, 34, 25, 12, 22, 11, 90]
    arr8 = [64, 34, 25, 12, 22, 11, 90]
    
    n = len(arr1)
    
    print("Original array:", arr1)
    print()
    
    bubble_sort(arr1, n)
    print("Bubble Sort:", arr1)
    
    selection_sort(arr2, n)
    print("Selection Sort:", arr2)
    
    insertion_sort(arr3, n)
    print("Insertion Sort:", arr3)
    
    merge_sort(arr4, 0, n - 1)
    print("Merge Sort:", arr4)
    
    quick_sort(arr5, 0, n - 1)
    print("Quick Sort:", arr5)
    
    heap_sort(arr6, n)
    print("Heap Sort:", arr6)
    
    counting_sort(arr7, n)
    print("Counting Sort:", arr7)
    
    radix_sort(arr8, n)
    print("Radix Sort:", arr8)


if __name__ == "__main__":
    main()
```
