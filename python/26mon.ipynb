```python
import random

class Result:
    @staticmethod
    def countSmallerElements(arr):
        """
        Counts how many elements are smaller than each element.
        Args:
            arr (list[int]): List of integers
        Returns:
            list[int]: Count of smaller elements for each index
        """
        n = len(arr)
        counts = []
        
        for i in range(n):
            count = 0
            for j in range(n):
                if arr[j] < arr[i]:
                    count += 1
            counts.append(count)
        
        return counts

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(0, 10) for _ in range(n)]
    print("Array:", arr)
    result = Result.countSmallerElements(arr)
    print("Result:", result)
```
