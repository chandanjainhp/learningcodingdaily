```python
import random

class Result:
    @staticmethod
    def countSmallerElements(arr):
        """
        Counts how many elements are smaller than each element.
        Args:
            arr (list[int]): List of integers
        Returns:
            list[int]: Count of smaller elements for each index
        """
        n = len(arr)
        counts = []
        
        for i in range(n):
            count = 0
            for j in range(n):
                if arr[j] < arr[i]:
                    count += 1
            counts.append(count)
        
        return counts

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(0, 10) for _ in range(n)]
    print("Array:", arr)
    result = Result.countSmallerElements(arr)
    print("Result:", result)
```

```python
import random

class Result:
    @staticmethod
    def findPairsWithSum(arr, target):
        """
        Finds all unique pairs with the given sum.
        
        Args:
            arr (list[int]): List of integers
            target (int): Target sum
        
        Returns:
            list[tuple[int, int]]: List of unique pairs
        """
        seen = set()
        pairs = set()
        
        for num in arr:
            complement = target - num
            
            # If complement exists in seen, we found a pair
            if complement in seen:
                # Store pair in sorted order to avoid duplicates like (2,3) and (3,2)
                pair = tuple(sorted([num, complement]))
                pairs.add(pair)
            
            seen.add(num)
        
        return sorted(list(pairs))

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(-10, 10) for _ in range(n)]
    target = random.randint(-5, 10)
    
    print("Array:", arr)
    print("Target:", target)
    result = Result.findPairsWithSum(arr, target)
    print("Pairs:", result)
```

```python
def productExceptSelf(nums):
    n = len(nums)
    answer = [1] * n
    
    # Left pass: answer[i] contains product of all elements to the left
    left_product = 1
    for i in range(n):
        answer[i] = left_product
        left_product *= nums[i]
    
    # Right pass: multiply by product of all elements to the right
    right_product = 1
    for i in range(n - 1, -1, -1):
        answer[i] *= right_product
        right_product *= nums[i]
    
    return answer


nums = [1, 2, 3, 4]
pro =productExceptSelf(nums)
print(pro)
```

```python
import random

class Result:
    @staticmethod
    def longestSubarrayAtMostKDistinct(arr, k):
        """
        Finds the length of the longest subarray with at most k distinct elements.
        
        Args:
            arr (list[int]): List of integers
            k (int): Maximum number of distinct elements allowed
        
        Returns:
            int: Length of the longest valid subarray
        """
        if not arr or k == 0:
            return 0
        
        # Dictionary to track frequency of elements in current window
        freq = {}
        left = 0
        max_length = 0
        
        # Expand window with right pointer
        for right in range(len(arr)):
            # Add current element to window
            freq[arr[right]] = freq.get(arr[right], 0) + 1
            
            # Shrink window while we have more than k distinct elements
            while len(freq) > k:
                freq[arr[left]] -= 1
                if freq[arr[left]] == 0:
                    del freq[arr[left]]
                left += 1
            
            # Update maximum length
            max_length = max(max_length, right - left + 1)
        
        return max_length

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    arr = [random.randint(1, 5) for _ in range(n)]
    k = random.randint(1, 3)
    
    print("Array:", arr)
    print("k:", k)
    result = Result.longestSubarrayAtMostKDistinct(arr, k)
    print("Longest Length:", result)
```
