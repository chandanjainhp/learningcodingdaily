```python
# ============================================
# 1. LINEAR SEARCH - O(n)
# ============================================
def linear_search(arr, n, key):
    """Search sequentially through array"""
    for i in range(n):
        if arr[i] == key:
            return i  # Return index if found
    return -1  # Return -1 if not found


# ============================================
# 2. BINARY SEARCH - O(log n) - Array must be sorted
# ============================================
def binary_search(arr, n, key):
    """Divide and conquer search on sorted array"""
    left = 0
    right = n - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        
        # Check if key is present at mid
        if arr[mid] == key:
            return mid
        
        # If key is greater, ignore left half
        elif arr[mid] < key:
            left = mid + 1
        
        # If key is smaller, ignore right half
        else:
            right = mid - 1
    
    return -1  # Element not found


# ============================================
# 3. BINARY SEARCH (Recursive) - O(log n)
# ============================================
def binary_search_recursive(arr, left, right, key):
    """Recursive version of binary search"""
    if right >= left:
        mid = left + (right - left) // 2
        
        # If element is present at middle
        if arr[mid] == key:
            return mid
        
        # If element is smaller than mid, search left half
        elif arr[mid] > key:
            return binary_search_recursive(arr, left, mid - 1, key)
        
        # Else search right half
        else:
            return binary_search_recursive(arr, mid + 1, right, key)
    
    else:
        return -1  # Element not present


# ============================================
# 4. JUMP SEARCH - O(√n) - Array must be sorted
# ============================================
def jump_search(arr, n, key):
    """Jump ahead by fixed steps, then linear search"""
    import math
    
    # Finding block size to jump
    step = int(math.sqrt(n))
    prev = 0
    
    # Jump to find block where element may be present
    while arr[min(step, n) - 1] < key:
        prev = step
        step = step + int(math.sqrt(n))
        
        # If we reached end of array
        if prev >= n:
            return -1
    
    # Linear search in identified block
    while arr[prev] < key:
        prev = prev + 1
        
        # If we reached next block or end of array
        if prev == min(step, n):
            return -1
    
    # If element is found
    if arr[prev] == key:
        return prev
    
    return -1


# ============================================
# 5. INTERPOLATION SEARCH - O(log log n) for uniform distribution
# ============================================
def interpolation_search(arr, n, key):
    """Better than binary for uniformly distributed sorted arrays"""
    left = 0
    right = n - 1
    
    while left <= right and key >= arr[left] and key <= arr[right]:
        # If array has only one element
        if left == right:
            if arr[left] == key:
                return left
            return -1
        
        # Probing the position with keeping uniform distribution in mind
        pos = left + int(((right - left) / (arr[right] - arr[left])) * (key - arr[left]))
        
        # Condition of target found
        if arr[pos] == key:
            return pos
        
        # If key is larger, key is in upper part
        if arr[pos] < key:
            left = pos + 1
        
        # If key is smaller, key is in lower part
        else:
            right = pos - 1
    
    return -1


# ============================================
# 6. EXPONENTIAL SEARCH - O(log n)
# ============================================
def exponential_search(arr, n, key):
    """Find range by exponential jumps, then binary search"""
    # If key is at first position
    if arr[0] == key:
        return 0
    
    # Find range for binary search by repeated doubling
    i = 1
    while i < n and arr[i] <= key:
        i = i * 2
    
    # Call binary search for found range
    return binary_search_recursive(arr, i // 2, min(i, n - 1), key)


# ============================================
# 7. TERNARY SEARCH - O(log₃ n) - Array must be sorted
# ============================================
def ternary_search(arr, left, right, key):
    """Divide array into three parts"""
    if right >= left:
        # Find mid1 and mid2
        mid1 = left + (right - left) // 3
        mid2 = right - (right - left) // 3
        
        # Check if key is at any mid
        if arr[mid1] == key:
            return mid1
        if arr[mid2] == key:
            return mid2
        
        # Since key is not present at mid, check in which region it is
        if key < arr[mid1]:
            # Key lies between left and mid1
            return ternary_search(arr, left, mid1 - 1, key)
        elif key > arr[mid2]:
            # Key lies between mid2 and right
            return ternary_search(arr, mid2 + 1, right, key)
        else:
            # Key lies between mid1 and mid2
            return ternary_search(arr, mid1 + 1, mid2 - 1, key)
    
    return -1


# ============================================
# 8. FIBONACCI SEARCH - O(log n)
# ============================================
def fibonacci_search(arr, n, key):
    """Uses Fibonacci numbers to divide array"""
    # Initialize fibonacci numbers
    fib_m_minus_2 = 0  # (m-2)'th Fibonacci number
    fib_m_minus_1 = 1  # (m-1)'th Fibonacci number
    fib_m = fib_m_minus_1 + fib_m_minus_2  # m'th Fibonacci
    
    # fib_m is going to store the smallest Fibonacci number >= n
    while fib_m < n:
        fib_m_minus_2 = fib_m_minus_1
        fib_m_minus_1 = fib_m
        fib_m = fib_m_minus_1 + fib_m_minus_2
    
    # Marks the eliminated range from front
    offset = -1
    
    # While there are elements to be inspected
    while fib_m > 1:
        # Check if fib_m_minus_2 is a valid location
        i = min(offset + fib_m_minus_2, n - 1)
        
        # If key is greater than value at index fib_m_minus_2, cut the subarray from offset to i
        if arr[i] < key:
            fib_m = fib_m_minus_1
            fib_m_minus_1 = fib_m_minus_2
            fib_m_minus_2 = fib_m - fib_m_minus_1
            offset = i
        
        # If key is less than value at index fib_m_minus_2, cut the subarray after i+1
        elif arr[i] > key:
            fib_m = fib_m_minus_2
            fib_m_minus_1 = fib_m_minus_1 - fib_m_minus_2
            fib_m_minus_2 = fib_m - fib_m_minus_1
        
        # Element found
        else:
            return i
    
    # Comparing the last element with key
    if fib_m_minus_1 and offset + 1 < n and arr[offset + 1] == key:
        return offset + 1
    
    return -1


# ============================================
# 9. SENTINEL LINEAR SEARCH - O(n) but fewer comparisons
# ============================================
def sentinel_linear_search(arr, n, key):
    """Linear search with sentinel to reduce comparisons"""
    # Last element of array
    last = arr[n - 1]
    
    # Element to be searched is placed at the last index
    arr[n - 1] = key
    i = 0
    
    # One comparison is removed as last element is key
    while arr[i] != key:
        i = i + 1
    
    # Put the last element back
    arr[n - 1] = last
    
    # Check if element found or we reached the sentinel
    if i < n - 1 or arr[n - 1] == key:
        return i
    else:
        return -1


# ============================================
# 10. META BINARY SEARCH (One-sided Binary Search)
# ============================================
def meta_binary_search(arr, n, key):
    """Binary search using bit manipulation"""
    # Set the left and right bounds
    lg = 0
    temp = n - 1
    
    # Calculate log2(n)
    while temp > 0:
        lg = lg + 1
        temp = temp >> 1  # Right shift (divide by 2)
    
    pos = 0
    
    # Iterate through each bit
    for i in range(lg - 1, -1, -1):
        if arr[pos] == key:
            return pos
        
        # Check if moving to next position is valid
        new_pos = pos | (1 << i)  # Set i-th bit
        
        # If valid position and value is less than or equal to key
        if new_pos < n and arr[new_pos] <= key:
            pos = new_pos
    
    # Check if element found
    if arr[pos] == key:
        return pos
    else:
        return -1


# ============================================
# MAIN FUNCTION - Test all search algorithms
# ============================================
def main():
    # Test arrays
    arr_unsorted = [64, 34, 25, 12, 22, 11, 90, 88, 45, 50]
    arr_sorted = [11, 12, 22, 25, 34, 45, 50, 64, 88, 90]
    
    n_unsorted = len(arr_unsorted)
    n_sorted = len(arr_sorted)
    
    key = 45
    
    print("Unsorted Array:", arr_unsorted)
    print("Sorted Array:", arr_sorted)
    print("Searching for:", key)
    print()
    
    # Linear Search (works on unsorted)
    result = linear_search(arr_unsorted, n_unsorted, key)
    print("1. Linear Search: Index =", result)
    
    # Sentinel Linear Search
    arr_temp = arr_unsorted.copy()
    result = sentinel_linear_search(arr_temp, n_unsorted, key)
    print("2. Sentinel Linear Search: Index =", result)
    
    # Binary Search (needs sorted array)
    result = binary_search(arr_sorted, n_sorted, key)
    print("3. Binary Search: Index =", result)
    
    # Binary Search Recursive
    result = binary_search_recursive(arr_sorted, 0, n_sorted - 1, key)
    print("4. Binary Search (Recursive): Index =", result)
    
    # Jump Search
    result = jump_search(arr_sorted, n_sorted, key)
    print("5. Jump Search: Index =", result)
    
    # Interpolation Search
    result = interpolation_search(arr_sorted, n_sorted, key)
    print("6. Interpolation Search: Index =", result)
    
    # Exponential Search
    result = exponential_search(arr_sorted, n_sorted, key)
    print("7. Exponential Search: Index =", result)
    
    # Ternary Search
    result = ternary_search(arr_sorted, 0, n_sorted - 1, key)
    print("8. Ternary Search: Index =", result)
    
    # Fibonacci Search
    result = fibonacci_search(arr_sorted, n_sorted, key)
    print("9. Fibonacci Search: Index =", result)
    
    # Meta Binary Search
    result = meta_binary_search(arr_sorted, n_sorted, key)
    print("10. Meta Binary Search: Index =", result)


if __name__ == "__main__":
    main()
```
