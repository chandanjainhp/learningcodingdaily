```python
import random

class Result:
    @staticmethod
    def countActiveStudents(status):
        """
        Counts the number of active students.

        Args:
            status (list[int]): List where 1 represents active and 0 represents inactive students

        Returns:
            int: Total number of active students
        """
        count = 0
        for student_status in status:
            if student_status == 1:
                count += 1
        return count

if __name__ == "__main__":
    # Random test data generation (for local testing)
    n = random.randint(5, 15)
    status = [random.randint(0, 1) for _ in range(n)]

    print("Number of Students:", n)
    print("Student Status List:", status)

    result = Result.countActiveStudents(status)
    print("Active Students Count:", result)
```

3️⃣ Inheritance ⭐⭐⭐⭐⭐ (Must Know)

Create a base class Person with attributes name and age.
Create a derived class Student that adds an attribute student_id.

Create a class Vehicle with a method start().
Inherit it in a class Bike and call the start() method using a Bike object.

Create a parent class Employee and a child class Manager.
Add a method in the parent and demonstrate that the child class can access it.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id

s = Student("Alice", 20, "S12345")
print(s.name)        # Alice
print(s.age)         # 20
print(s.student_id)  # S12345
```

```python
class Employee:
    def __init__(self, emp_id, salary):
        self.emp_id = emp_id
        self.salary = salary


class Manager(Employee):
    def __init__(self, emp_id, salary, dateofjoin):
        super().__init__(emp_id, salary)
        self.dateofjoin = dateofjoin
            
    def __str__(self):
        return f"This is employee {self.emp_id}, salary {self.salary}, joined on {self.dateofjoin}"


manager = Manager("3434", "89k", "20-02-2026")
print(manager)
```

```python
'''
Docstring for 06_inheritance_composition
Composition is a design principle in object-oriented 
programming where one class (the "composite" or "container" class) 
contains an instance of another class (the "component" class) as an attribute.
This allows the composite class to delegate behavior to the component, promoting code reuse and flexibility.
Unlike inheritance (where a subclass inherits properties from a parent),
composition avoids tight coupling and enables runtime changes to the component. 
For example, a Car class might compose an Engine object, allowing different engine types 
to be swapped without changing the car's structure
'''

# Base class for chai types, demonstrating inheritance
class BaseChai:
    def __init__(self, type_):
        self.type = type_

    def prepare(self):
        print(f"Preparing {self.type} chai....")

# Subclass of BaseChai, adds spice functionality via inheritance
class MasalaChai(BaseChai):
    def add_spices(self):
        print("Adding cardamom, ginger, cloves.")

# Class demonstrating composition: contains a chai object
class ChaiShop:
    chai_cls = BaseChai  # Class attribute for the chai type (can be overridden)

    def __init__(self):
        # Composition: ChaiShop creates and holds a chai instance
        self.chai = self.chai_cls("Regular")

    def serve(self):
        print(f"Serving {self.chai.type} chai in the shop")
        self.chai.prepare()

# Subclass of ChaiShop, overrides chai_cls for different behavior
class FancyChaiShop(ChaiShop):
    chai_cls = MasalaChai  # Override to use MasalaChai instead

# Create instances and demonstrate the behavior
shop = ChaiShop()
fancy = FancyChaiShop()
shop.serve()
fancy.serve()
fancy.chai.add_spices()
```
